# Week 3 Tutorial 3 Documentation

## Program Name
Student Average Calculator

## Scenario
This program calculates the average mark for three quizzes and checks if the student passes or fails.

## Inputs
- Quiz 1 mark
- Quiz 2 mark
- Quiz 3 mark

## Process
- Enter three quiz marks.
- Add the three marks.
- Divide the total by 3.
- Check if the average is 50 or above.

## Output
- Average mark
- Pass or Fail
