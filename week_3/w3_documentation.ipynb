# Week 3 Tutorial 3 Documentation

## Program Name
Student Average Calculator

## Scenario
This program calculates the average mark for three quizzes and checks if the student passes or fails.

## Inputs
- Quiz 1 mark
- Quiz 2 mark
- Quiz 3 mark

## Process
- Enter three quiz marks.
- Add the three marks.
- Divide the total by 3.
- Check if the average is 50 or above.

## Output
- Average mark
- Pass or Fail

## 2. Design the Algorithm

### 2.1 Pseudocode
START

Set choice to y

WHILE choice is y

Input Quiz 1 mark
Input Quiz 2 mark
Input Quiz 3 mark

Calculate average = (Quiz 1 + Quiz 2 + Quiz 3) / 3

Display average

IF average >= 50 THEN
Display Pass
ELSE
Display Fail

Ask user if they want to continue

END WHILE

Display Program Ended

END

### 2.2 Algorithm Structure

#### 2.2.1 Sequence
The program follows steps in order: input marks, calculate average, display result.

#### 2.2.2 Selection
The program uses IF and ELSE to decide whether the student passes or fails.

#### 2.2.3 Iteration
The program uses WHILE to allow another student's marks to be entered.
